이 관습은 선형대수학(Linear Algebra)의 수학적 표기법에서 유래되었습니다. 프로그래밍 언어의 문법적 요구사항이 아니라, 데이터를 다루는 **수학적 차원(Dimensionality)** 을 시각적으로 구분하기 위한 약속입니다.

### 핵심 이유: 행렬(Matrix) vs 벡터(Vector)

1.  **대문자 $\mathbf{X}$ (행렬/Matrix):**
    *   머신러닝에서 입력 데이터는 보통 **2차원 배열**입니다. (행: 샘플 수, 열: 특성 수)
    *   수학적으로 2차원 이상의 배열은 **행렬(Matrix)** 로 정의하며, 행렬은 관습적으로 **대문자**로 표기합니다.
    *   따라서 `X`를 보면 "이 변수는 행과 열이 있는 2차원 구조의 데이터셋(Feature Matrix)이구나"라고 즉시 알 수 있습니다.

2.  **소문자 $\mathbf{y}$ (벡터/Vector):**
    *   정답(타겟) 데이터는 보통 각 샘플당 하나의 값만 가지는 **1차원 배열**입니다.
    *   수학적으로 1차원 배열은 **벡터(Vector)** 로 정의하며, 벡터는 관습적으로 **소문자**로 표기합니다.
    *   따라서 `y`를 보면 "이 변수는 정답들이 나열된 1차원 리스트(Target Vector)구나"라고 판단합니다.

### 프로그래머 관점에서의 이점

이 표기법을 따르면 코드의 가독성과 유지보수 효율이 올라갑니다.

*   **차원 오류 방지:** `X`와 `y`를 구분함으로써, 2차원 행렬이 들어가야 할 곳에 1차원 벡터를 넣거나 그 반대의 실수를 하는 것을 시각적으로 빠르게 인지할 수 있습니다. (예: `X.shape`는 `(n_samples, n_features)`이고, `y.shape`는 `(n_samples,)`임을 기대함)
*   **커뮤니티 표준:** scikit-learn뿐만 아니라 TensorFlow, PyTorch, NumPy 기반의 거의 모든 ML 라이브러리와 논문에서 이 표기법을 사용합니다. 표준을 따르면 다른 개발자의 코드를 읽거나 공식 문서를 이해하는 속도가 빨라집니다.

### 요약
| 변수 | 표기 | 수학적 의미 | 데이터 구조 | 차원 |
| :--- | :--- | :--- | :--- | :--- |
| **특성(Features)** | `X` | 행렬 (Matrix) | 2D Array | $\mathbb{R}^{n \times m}$ |
| **타겟(Target)** | `y` | 벡터 (Vector) | 1D Array | $\mathbb{R}^{n}$ |

scikit-learn에서 **`X`는 관례상 입력 데이터(특성)를 나타내며, 대문자인 이유는 보통 행렬로 표현되기 때문**입니다. `X`의 각 행은 샘플 하나, 각 열은 특성 하나를 뜻하고, 일반적인 모양은 `(n_samples, n_features)`입니다. 예를 들어 질문의 `X`는 샘플 2개와 특성 3개이므로 모양이 `(2, 3)`입니다. [scikit-learn.org](https://scikit-learn.org/stable/glossary.html#term-X)

반면 `y`는 보통 예측하려는 정답·레이블을 나타내며, 샘플마다 하나의 값이 있는 1차원 벡터인 경우가 많습니다. 그래서 흔히 `X`는 대문자, `y`는 소문자로 씁니다. 여러 출력값을 행렬로 나타내는 경우에는 `Y`라고 쓰기도 합니다. [scikit-learn.org](https://scikit-learn.org/stable/glossary.html#term-X)

이건 **이름을 정하는 관례이지, 파이썬이나 scikit-learn의 강제 규칙은 아닙니다.** `features`처럼 다른 변수명을 써도 되지만, `X`와 `y`를 사용하면 scikit-learn 문서와 예제의 표기법을 그대로 따라갈 수 있어 읽기 편합니다.

## Fitting and predicting: estimator basics

## Transformers and pre-processors

```python
from sklearn import set_config
from sklearn.ensemble import RandomForestClassifier
from sklearn.preprocessing import StandardScaler

set_config(display="text")

# https://scikit-learn.org/stable/getting_started.html#fitting-and-predicting-estimator-basics
clf = RandomForestClassifier(random_state=0)
X = [[1, 2, 3], [11, 12, 13]]  # 2 samples, 3 features
y = [0, 1]  # classes of each sample


print(clf.fit(X, y))

print(clf.predict(X))  # predict classes of the training data
print(clf.predict([[4, 5, 6], [14, 15, 16]]))  # predict classes of new data

# https://scikit-learn.org/stable/getting_started.html#transformers-and-pre-processors
"""scale data according to computed scaling values"""
print(StandardScaler().fit([[0, 15], [1, -10]]).transform([[0, 15], [1, -10]]))
```

## Pipelines: chaining pre-processors and estimators

```python
# https://scikit-learn.org/stable/getting_started.html#pipelines-chaining-pre-processors-and-estimators

from sklearn.datasets import load_iris
from sklearn.linear_model import LogisticRegression
from sklearn.metrics import accuracy_score
from sklearn.model_selection import train_test_split
from sklearn.pipeline import make_pipeline
from sklearn.preprocessing import StandardScaler

# create a pipeline object
pipe = make_pipeline(StandardScaler(), LogisticRegression())

# load the iris dataset and split it into train and test sets
X, y = load_iris(return_X_y=True)
X_train, X_test, y_train, y_test = train_test_split(X, y, random_state=0)

# fit the whole pipeline
print(pipe.fit(X_train, y_train))

# we can now use it like any other estimator
print(accuracy_score(pipe.predict(X_test), y_test))
```

## Model evaluation

```python
# https://scikit-learn.org/stable/getting_started.html#model-evaluation

from sklearn.datasets import make_regression
from sklearn.linear_model import LinearRegression
from sklearn.model_selection import cross_validate

X, y = make_regression(n_samples=1000, random_state=0)
lr = LinearRegression()

result = cross_validate(lr, X, y)  # defaults to 5-fold CV
print(result["test_score"])  # r_squared score is high because dataset is easy
```

## Automatic parameter searches

```python
# https://scikit-learn.org/stable/getting_started.html#automatic-parameter-searches

from scipy.stats import randint
from sklearn.datasets import make_regression
from sklearn.ensemble import RandomForestRegressor
from sklearn.model_selection import RandomizedSearchCV, train_test_split

# create a synthetic dataset
X, y = make_regression(n_samples=20640, n_features=8, noise=0.1, random_state=0)

X_train, X_test, y_train, y_test = train_test_split(X, y, random_state=0)

# define the parameter space that will be searched over
param_distributions = {"n_estimators": randint(1, 5), "max_depth": randint(5, 10)}

# now create a searchCV object and fit it to the data
search = RandomizedSearchCV(
    estimator=RandomForestRegressor(random_state=0),
    n_iter=5,
    param_distributions=param_distributions,
    random_state=0,
)
print(search.fit(X_train, y_train))
print(search.best_params_)

# the search object now acts like a normal random forest estimator
# with max_depth=9 and n_estimators=4
print(search.score(X_test, y_test))
```
