```python
import matplotlib.pyplot as plt
import torch
from torch import nn, optim
from torchvision import datasets, transforms

device = torch.device("cuda" if torch.cuda.is_available() else "cpu")
print(device)


class ConvAutoencoder(nn.Module):
    def __init__(self):
        super().__init__()

        # Encoder layers
        self.encoder = nn.Sequential(
            nn.Conv2d(1, 16, kernel_size=3, stride=2, padding=1),  # (16, 14, 14)
            nn.ReLU(),

            nn.Conv2d(16, 32, kernel_size=3, stride=2, padding=1),  # (32, 7, 7)
            nn.ReLU(),

            nn.Conv2d(32, 64, kernel_size=7),  # (64, 1, 1) - Latent space            
        )

        # Decoder layers
        self.decoder = nn.Sequential(
            nn.ConvTranspose2d(64, 32, kernel_size=7),  # (32, 7, 7)
            nn.ReLU(),
            
            nn.ConvTranspose2d(32, 16, kernel_size=3, stride=2, padding=1, output_padding=1),  # (16, 14, 14)
            nn.ReLU(),

            nn.ConvTranspose2d(16, 1, kernel_size=3, stride=2, padding=1, output_padding=1),  # (1, 28, 28)
            nn.Sigmoid(),  # Output pixel values are between 0 and 1
        )

    def forward(self, x):
        x = self.encoder(x)
        x = self.decoder(x)
        return x


# 데이터 로딩
transform = transforms.Compose(
    [
        transforms.ToTensor(),  # 텐서로 변환
        # transforms.Normalize((0.5, ), (0.5,))  # sigmoid에서 0.5 정규화는 부적절
    ]
)
train_dataset = datasets.MNIST(
    root="./data", train=True, transform=transform, download=True
)
train_loader = torch.utils.data.DataLoader(
    train_dataset,
    batch_size=64,
    shuffle=True,
    pin_memory=(device.type == "cuda"),
    # DataLoader의 셔플용 샘플러는 인덱스와 난수 생성기를 CPU에서 사용합니다.
    generator=torch.Generator(device="cpu"),
)


# Initialize the Convolutional Autoencoder, Loss function, and Optimizer
convolution_model = ConvAutoencoder().to(device)  # 모델과 입력 텐서가 모두 같은 장치(CPU 또는 CUDA)에 있어야 한다
convolution_criterion = nn.MSELoss()
convolution_optimizer = optim.Adam(convolution_model.parameters(), lr=0.001)
# conv_optimizer = optim.AdamW(conv_model.parameters(), lr=0.001)

# Training loop for Convolutional Autoencoder 
num_epochs = 20  # Can adjust this

print("\nStarting training for Convolutional Autoencoder...")

for epoch in range(num_epochs):
    for data in train_loader:
        img, _ = data  # label을 쓰지 않는 비지도학습

        img = img.to(
            device,
            non_blocking=(device.type == "cuda")
        )

        # No need to flatten for Conv layers
        reconstruct = convolution_model(img)
        loss = convolution_criterion(reconstruct, img)

        convolution_optimizer.zero_grad()
        loss.backward()
        convolution_optimizer.step()

        print(f"Epoch [{epoch + 1}/{num_epochs}], Loss: {loss.item():.4f}")


# Visualize reconstruction results for Convolutional Autoencoder
convolution_model.eval()
with torch.no_grad():
    # dataiter = iter(test_loader)
    dataiter = iter(train_loader)
    images, _ = next(dataiter)
    images = images.to(
        device,
        non_blocking=(device.type == "cuda")
    )
    reconstructed_images = convolution_model(images)

    n = 10
    plt.figure(figsize=(20, 4))
    for i in range(n):
        # Original Images
        ax = plt.subplot(2, n, i+1)
        plt.imshow(images[i].squeeze().cpu().numpy(), cmap="gray")  # Squeeze the channel dimension
        ax.get_xaxis().set_visible(False)
        ax.get_yaxis().set_visible(False)

        # Reconstructed Images
        ax = plt.subplot(2, n, i + 1 + n)
        plt.imshow(reconstructed_images[i].squeeze().cpu().numpy(), cmap="gray")  # 시각화할 때는 CUDA 텐서를 바로 .numpy()로 변환할 수 없으므로 CPU로 옮긴 뒤 변환
        ax.get_xaxis().set_visible(False)
        ax.get_yaxis().set_visible(False)

    plt.show()
```

```python
import matplotlib.pyplot as plt
import torch
from torch import nn, optim
from torchvision import datasets, transforms

device = torch.device("cuda" if torch.cuda.is_available() else "cpu")
print(device)


class VAE(nn.Module):
    def __init__(self, latent_dim):
        super().__init__()

        self.fully_connected1 = nn.Linear(28*28, 512)
        self.fully_connected2_mu = nn.Linear(512, latent_dim)
        self.fully_connected2_logvar = nn.Linear(512, latent_dim)
        self.fully_connected3 = nn.Linear(latent_dim, 512)
        self.fully_connected4 = nn.Linear(512, 28*28)

        self.relu = nn.ReLU()
        self.sigmoid = nn.Sigmoid()

    def encode(self, x):
        h1 = self.relu(self.fully_connected1(x))
        return self.fully_connected2_mu(h1), self.fully_connected2_logvar(h1)

    def reparameterize(self, mu, logvar):
        std = torch.exp(0.5 * logvar)
        eps = torch.randn_like(std)
        return mu + eps * std

    def decode(self, z):
        h3 = self.relu(self.fully_connected3(z))
        return self.sigmoid(self.fully_connected4(h3))

    def forward(self, x):
        mu, logvar = self.encode(x.view(-1, 28*28))
        z = self.reparameterize(mu, logvar)
        return self.decode(z), mu, logvar


# VAR Loss Function
def var_loss_function(reconstruct_x, x, mu, logvar):
    BCE = nn.functional.binary_cross_entropy(reconstruct_x, x.view(-1, 28*28), reduction="sum")
    KLD = -0.5 * torch.sum(1 + logvar - mu.pow(2) - logvar.exp())
    return BCE + KLD

# 데이터 로딩
transform = transforms.Compose(
    [
        transforms.ToTensor(),  # 텐서로 변환
        # transforms.Normalize((0.5, ), (0.5,))  # sigmoid에서 0.5 정규화는 부적절
    ]
)
train_dataset = datasets.MNIST(
    root="./data", train=True, transform=transform, download=True
)
train_loader = torch.utils.data.DataLoader(
    train_dataset,
    batch_size=64,
    shuffle=True,
    pin_memory=(device.type == "cuda"),
    # DataLoader의 셔플용 샘플러는 인덱스와 난수 생성기를 CPU에서 사용합니다.
    generator=torch.Generator(device="cpu"),
)


# Initialize VAE
latent_dim = 20  # You can adjust the latent dimension
vae_model = VAE(latent_dim).to(device)  # 모델과 입력 텐서가 모두 같은 장치(CPU 또는 CUDA)에 있어야 한다
vae_optimizer = optim.Adam(vae_model.parameters(), lr=0.001)
# vae_optimizer = optim.AdamW(vae_model.parameters(), lr=0.001)

# Training loop for VAE
num_epochs = 20  # Can adjust this

print("\nStarting training for Variational Autoencoder...")

for epoch in range(num_epochs):
    vae_model.train()
    train_loss = 0
    
    for batch_idx, (data, _) in enumerate(train_loader):
        data = data.to(vae_model.fully_connected1.weight.device)  # Ensure data is on the same device as the model

        vae_optimizer.zero_grad()
        reconstruct, mu, logvar = vae_model(data)

        loss = var_loss_function(reconstruct, data, mu, logvar)
        loss.backward()
        train_loss += loss.item()
        vae_optimizer.step()

    print(f"Epoch [{epoch + 1}/{num_epochs}], Loss: {train_loss/len(train_loader):.4f}")


# Visualize reconstruction results for VAE
vae_model.eval()
with torch.no_grad():
    dataiter = iter(train_loader)
    images, _ = next(dataiter)
    images = images.to(
        device,
        non_blocking=(device.type == "cuda")
    )
    reconstructed_images = vae_model(images)[0]
    reconstructed_images = reconstructed_images.view(-1, 1, 28, 28)

    n = 10
    plt.figure(figsize=(20, 4))
    for i in range(n):
        # Original Images
        ax = plt.subplot(2, n, i+1)
        plt.imshow(images[i].squeeze().cpu().numpy(), cmap="gray")  # Squeeze the channel dimension
        ax.get_xaxis().set_visible(False)
        ax.get_yaxis().set_visible(False)

        # Reconstructed Images
        ax = plt.subplot(2, n, i + 1 + n)
        plt.imshow(
            reconstructed_images[i].squeeze().cpu().numpy(), cmap="gray"
        )  # 시각화할 때는 CUDA 텐서를 바로 .numpy()로 변환할 수 없으므로 CPU로 옮긴 뒤 변환
        ax.get_xaxis().set_visible(False)
        ax.get_yaxis().set_visible(False)

    plt.show()
```

```python
import matplotlib.pyplot as plt
import numpy as np
import torch
from PIL import Image
from torch import optim
from torchvision import models, transforms
from torchvision.models import VGG19_Weights

device = torch.device("cuda" if torch.cuda.is_available() else "cpu")
print(device)

# RGBA·grayscale 이미지도 VGG 입력에 맞게 RGB로 변환
content_img = Image.open("./data/style-transfer/janelle.png").convert("RGB")
style_img = Image.open("./data/style-transfer/delaunay.jpg").convert("RGB")

print("content_img")
plt.imshow(content_img)
plt.show()
print("style_img")
plt.imshow(style_img)
plt.show()

# Preprocessing
img_transform = transforms.Compose(
    [
        transforms.Resize((400, 400)),
        transforms.ToTensor(),
        transforms.Normalize((0.485, 0.456, 0.406), (0.229, 0.224, 0.225)),
    ]
)

# content = img_transform(content_img)[:3,:,:].unsqueeze(0).to(device)
# style = img_transform(style_img)[:3, :, :].unsqueeze(0).to(device)

print(content_img.mode, style_img.mode)
content = img_transform(content_img).unsqueeze(0).to(device)
style = img_transform(style_img).unsqueeze(0).to(device)
print(content.shape, style.shape)  # 각각 [1, 3, 400, 400]이어야 함

# VGG19의 convolutional feature 부분만 사용하고 가중치는 고정
# vgg = models.vgg19(pretrained=True).features.to(device).eval()  # pretrained=True는 구식 인자입니다. VGG19_Weights.DEFAULT로 사전학습 가중치를 지정할 수 있습니다.

vgg = (
    models.vgg19(weights=VGG19_Weights.DEFAULT).features.to(device).eval()
)  # pretrained=True 대신 weights API 사용
# print(vgg)


for param in vgg.parameters():
    param.requires_grad_(False)  # 스타일, 특징 추출만 하고 학습은 하지 않음

# content는 이미 device에 있으므로, 최적화할 leaf tensor로 생성
target = content.clone().requires_grad_(True).to(device)
optimizer = optim.Adam([target], lr=0.003)

feature_layers = {
    "0": "conv1_1",
    "5": "conv2_1",
    "10": "conv3_1",
    "19": "conv4_1",
    "21": "conv4_2",
    "28": "conv5_1",
}


content_features = {}
img = content  # copy of content image
for name, layer in vgg._modules.items():
    img = layer(img)
    if name in feature_layers:
        content_features[feature_layers[name]] = img


style_features = {}
img = style  # copy of style image
for name, layer in vgg._modules.items():
    img = layer(img)
    if name in feature_layers:
        style_features[feature_layers[name]] = img


# a general function to extract the features
def get_features(image, model, layers=None):
    if layers is None:
        layers = {
            "0": "conv1_1",
            "5": "conv2_1",
            "10": "conv3_1",
            "19": "conv4_1",
            "21": "conv4_2",
            "28": "conv5_1",
        }

    features = {}
    x = image
    for name, layer in model._modules.items():
        x = layer(x)
        if name in layers:
            features[layers[name]] = x

    return features


def grammatrix(tensor):
    n, d, h, w = tensor.size()
    tensor = tensor.view(d, h * w)
    grammat = torch.mm(tensor, tensor.t())
    return grammat


style_gram_matrix = {
    layer: grammatrix(style_features[layer]) for layer in style_features
}


style_weights = {
    "conv1_1": 1.0,
    "conv2_1": 0.75,
    "conv3_1": 0.2,
    "conv4_1": 0.2,
    "conv5_1": 0.2,
}

content_weight = 1  # this is alpha in the final loss
style_weight = 1  # this is beta in the final loss


def im_convert(tensor):
    """Display a tensor as an image."""

    image = tensor.to("cpu").clone().detach()
    image = image.numpy().squeeze()
    image = image.transpose(1, 2, 0)
    image = image * np.array((0.229, 0.224, 0.225)) + np.array((0.485, 0.456, 0.406))
    image = np.clip(image, 0, 1)

    return image


show_img = 200
steps = 2000

for ii in range(1, steps + 1):
    target_features = get_features(target, vgg)
    content_loss = torch.mean(
        (target_features["conv4_2"] - content_features["conv4_2"]) ** 2
    )

    style_loss = 0
    for layer in style_weights:
        target_feature = target_features[layer]
        target_gram = grammatrix(target_feature)
        _, d, h, w = target_feature.shape
        style_gram = style_gram_matrix[layer]
        layer_style_loss = style_weights[layer] * torch.mean(
            (target_gram - style_gram) ** 2
        )
        style_loss += layer_style_loss / (d * h * w)

    total_loss = content_weight * content_loss + style_weight * style_loss

    optimizer.zero_grad()
    total_loss.backward()
    optimizer.step()

    if ii % show_img == 0:
        print(f"Step {ii}/{steps}, Total loss: {total_loss.item():.4f}")
        plt.imshow(im_convert(target))
        plt.grid(True)
        plt.show()


fig, (ax1, ax2) = plt.subplots(1, 2, figsize=(20, 10))
ax1.grid(False)
ax2.grid(False)
ax1.set_title("content image", fontsize="35", color="black")
ax2.set_title("target image", fontsize="35", color="black")
ax1.imshow(im_convert(content))
ax2.imshow(im_convert(target))
plt.show()
```
